# TASK 5 - MODEL EVALUATION CHECKLIST (CLASSIFICATION)

This is a classification problem (`cardio` is 0/1), so we follow the **[BOTH]** and **[CLASS]** items from the checklist. Only **one** advanced model is imported and used here: **AdaBoost**.

## 6. Try Advanced Models

### AdaBoost

```python
from sklearn.ensemble import AdaBoostClassifier

model_ada = AdaBoostClassifier(random_state=42)
model_ada.fit(X_train_scaled, y_train)
y_pred_ada = model_ada.predict(X_test_scaled)
```

## 1. Model Evaluation

Calculate Accuracy, Precision, Recall, F1-score for the Logistic Regression model (already trained earlier) and the AdaBoost model.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def evaluate(name, y_true, y_pred):
    acc = accuracy_score(y_true, y_pred)
    prec = precision_score(y_true, y_pred)
    rec = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    print(f"{name} -> Accuracy: {acc:.3f}, Precision: {prec:.3f}, Recall: {rec:.3f}, F1: {f1:.3f}")
    return acc, prec, rec, f1

logreg_metrics = evaluate("Logistic Regression", y_test, y_pred)
ada_metrics = evaluate("AdaBoost", y_test, y_pred_ada)
```

## 2. Check Overfitting / Underfitting

Compare Train score vs Test score for each model.

```python
def check_fit(name, model, X_train, y_train, X_test, y_test):
    train_score = model.score(X_train, y_train)
    test_score = model.score(X_test, y_test)
    gap = train_score - test_score
    print(f"{name} -> Train: {train_score:.3f}, Test: {test_score:.3f}, Gap: {gap:.3f}")
    if gap > 0.1:
        print("   -> Likely OVERFITTING (train much higher than test)")
    elif train_score < 0.6 and test_score < 0.6:
        print("   -> Likely UNDERFITTING (both scores low)")
    else:
        print("   -> Good fit (train and test scores are close)")

check_fit("Logistic Regression", model, X_train_scaled, y_train, X_test_scaled, y_test)
check_fit("AdaBoost", model_ada, X_train_scaled, y_train, X_test_scaled, y_test)
```

## 3. Cross-Validation (5-Fold)

Run 5-fold cross-validation on the training data for each model. Note the average score and the spread (std) across folds.

```python
from sklearn.model_selection import cross_val_score

def run_cv(name, estimator, X, y, cv=5):
    scores = cross_val_score(estimator, X, y, cv=cv)
    print(f"{name} -> CV scores: {scores}")
    print(f"{name} -> Mean: {scores.mean():.3f}, Std: {scores.std():.3f}")
    return scores

cv_logreg = run_cv("Logistic Regression", LogisticRegression(), X_train_scaled, y_train)
cv_ada = run_cv("AdaBoost", AdaBoostClassifier(random_state=42), X_train_scaled, y_train)
```

## 4. Compare All Models

Fill in the classification comparison table for each model tried.

```python
comparison_table = pd.DataFrame({
    "Model": ["Logistic Regression", "AdaBoost"],
    "Accuracy": [logreg_metrics[0], ada_metrics[0]],
    "Precision": [logreg_metrics[1], ada_metrics[1]],
    "Recall": [logreg_metrics[2], ada_metrics[2]],
    "F1-score": [logreg_metrics[3], ada_metrics[3]],
    "CV Mean": [cv_logreg.mean(), cv_ada.mean()],
    "CV Std": [cv_logreg.std(), cv_ada.std()],
})

comparison_table
```

```python
# Pick the model with the best score AND a stable (low-spread) cross-validation result
best_row = comparison_table.sort_values(by=["Accuracy", "CV Std"], ascending=[False, True]).iloc[0]
print("Best model based on Accuracy and CV stability:", best_row["Model"])
```

## 5. Hyperparameter Tuning

Use GridSearchCV on AdaBoost, the single advanced model imported above, and try different values of `n_estimators` and `learning_rate`.

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "n_estimators": [50, 100],
    "learning_rate": [0.5, 1.0]
}

grid_search = GridSearchCV(AdaBoostClassifier(random_state=42), param_grid, cv=3)
grid_search.fit(X_train_scaled, y_train)

print("Best combination found:", grid_search.best_params_)
print("Best CV score:", round(grid_search.best_score_, 3))
```

### Re-test the tuned model on the test set

```python
tuned_model = grid_search.best_estimator_
y_pred_tuned = tuned_model.predict(X_test_scaled)
tuned_accuracy = accuracy_score(y_test, y_pred_tuned)

print("AdaBoost accuracy BEFORE tuning:", round(ada_metrics[0], 3))
print("AdaBoost accuracy AFTER tuning:", round(tuned_accuracy, 3))
```
